```python
df = pd.read_csv('/content/Dataset  (1).csv')

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print(df.columns.tolist())
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
df['Cuisines'] = df['Cuisines'].fillna('Unknown')

print("Missing Cuisines after cleaning:",
      df['Cuisines'].isnull().sum())
```

```python
recommendation_data = df[
    ['Restaurant ID', 'Restaurant Name', 'Cuisines', 'Price range']
].copy()

recommendation_data.head()
```

```python
user_cuisine = "North Indian"
user_price_range = 2

print("Preferred cuisine:", user_cuisine)
print("Preferred price range:", user_price_range)
```

```python
recommended = recommendation_data[
    (recommendation_data['Cuisines'].str.contains(
        user_cuisine, case=False, na=False
    )) &
    (recommendation_data['Price range'] == user_price_range)
]

print("Number of matching restaurants:", len(recommended))

recommended.head(10)
```

```python
recommended[['Restaurant Name', 'Cuisines', 'Price range']].head(10)
```

```python
cuisine_text = recommendation_data['Cuisines'].fillna('Unknown')

print(cuisine_text.head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

cuisine_matrix = vectorizer.fit_transform(cuisine_text)

print("Cuisine matrix shape:", cuisine_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(cuisine_matrix)

print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
restaurant_index = 0

similar_scores = similarity_matrix[restaurant_index]

print("Selected restaurant:")
print(recommendation_data.iloc[restaurant_index]['Restaurant Name'])

print("\nTop similar restaurant indexes:")
print(similar_scores.argsort()[-6:][::-1])
```

```python
similar_indexes = similar_scores.argsort()[-6:][::-1]

similar_indexes = [i for i in similar_indexes if i != restaurant_index]

recommendations = recommendation_data.iloc[similar_indexes]

recommendations[['Restaurant Name', 'Cuisines', 'Price range']]
```

```python
print("Available price ranges:")
print(sorted(recommendation_data['Price range'].unique()))
```

```python
print("User preferred cuisine:", user_cuisine)
print("User preferred price range:", user_price_range)
```

```python
price_matched = recommendation_data[
    recommendation_data['Price range'] == user_price_range
].copy()

print("Restaurants matching the preferred price range:",
      len(price_matched))
```

```python
price_matches = recommendation_data[
    recommendation_data['Price range'] == user_price_range
].copy()

print("Restaurants matching the preferred price range:", len(price_matches))
```

```python
price_match_indexes = price_matches.index

print("Number of matching indexes:", len(price_match_indexes))
```

```python
price_similarity_scores = similarity_matrix[restaurant_index, price_match_indexes]

print("Number of similarity scores:", len(price_similarity_scores))
```

```python
top_positions = price_similarity_scores.argsort()[::-1][:10]

print("Top 10 similarity positions:")
print(top_positions)
```

```python
top_restaurant_indexes = price_match_indexes[top_positions]

print("Top restaurant indexes:")
print(top_restaurant_indexes)
```

```python
final_recommendations = recommendation_data.loc[
    top_restaurant_indexes,
    ['Restaurant Name', 'Cuisines', 'Price range']
]

final_recommendations
```

```python
final_matches = cuisine_matches[
    cuisine_matches['Price range'] == user_price_range
].copy()

print("Restaurants matching cuisine and price range:", len(final_matches))
```

```python
match_indexes = final_matches.index

match_similarity = similarity_matrix[
    restaurant_index, match_indexes
]

top_matches = match_similarity.argsort()[::-1][:10]

print("Top 10 matching positions:")
print(top_matches)
```

```python
top_match_indexes = match_indexes[top_matches]

print("Top restaurant indexes:")
print(top_match_indexes)
```

```python
final_recommendations = recommendation_data.loc[
    top_match_indexes,
    ['Restaurant Name', 'Cuisines', 'Price range']
]

final_recommendations
```

```python
final_recommendations = final_recommendations.drop_duplicates(
    subset='Restaurant Name'
)

final_recommendations
```
